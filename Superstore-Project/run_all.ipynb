```python
import subprocess
import os
import sys

def run_all():
    print("开始一键复现 Superstore 商业分析项目")

    scripts = [
        "01_load.ipynb",
        "02_clean.ipynb",
        "03_analysis_models.ipynb",
        "04_visuals.ipynb"
    ]

    scripts_dir = "scripts"
    
    if not os.path.exists(scripts_dir):
        print(f"找不到 {scripts_dir} 文件夹，请确保在项目根目录下运行此脚本。")
        sys.exit(1)

    for script in scripts:
        file_path = os.path.join(scripts_dir, script)
        
        if not os.path.exists(file_path):
            print(f"找不到文件 {file_path}，跳过该步骤。")
            continue

        print(f"\n正在执行：{script} ...")

        cmd = [
            "jupyter", "nbconvert", 
            "--to", "notebook", 
            "--execute", 
            "--inplace", 
            file_path
        ]
        
        try:
            subprocess.run(cmd, check=True, stdout=subprocess.DEVNULL, stderr=subprocess.PIPE)
            print(f"{script} 执行成功！")
        except subprocess.CalledProcessError as e:
            print(f"{script} 执行失败！")
            print(f"错误信息: {e.stderr.decode('utf-8', errors='ignore')}")
            sys.exit(1)

    print("全部数据处理、模型计算与绘图流程已执行完毕")
    print("前往 outputs/figures/ 查看图表，outputs/tables/ 查看数据表。")
    print("若需查看交互式大屏，请手动运行05_dashboard.ipynb")

if __name__ == "__main__":
    run_all()
```
